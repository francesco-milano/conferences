```python
# Analysis of Delta table bronze.orders in the current lakehouse

table_name = "bronze.orders"

from delta.tables import DeltaTable
from pyspark.sql.functions import col

# 1) Table properties (e.g., CDF enabled)
print("=== 1) \"DESCRIBE DETAIL\" of the table ===")
df_detail = spark.sql(f"DESCRIBE DETAIL {table_name}")
display(df_detail)

props_row = df_detail.select("properties").collect()[0]
props = props_row["properties"] if props_row["properties"] is not None else {}

cdf_enabled = props.get("delta.enableChangeDataFeed", "not set")
print(f"\nCDF enabled? {cdf_enabled}")
```

```python
# 2) Available Delta history (list of versions)

from delta.tables import DeltaTable

print("=== 2) History Delta (version list) ===")

delta_tbl = DeltaTable.forName(spark, "bronze.orders")

history_df = delta_tbl.history()
display(history_df)
```

```python
# 3) Snapshot of a specific Delta version

version_to_check = 0

print(f"=== 3) Snapshot of the table at version {version_to_check}")

df_version = spark.sql(f"SELECT * FROM bronze.orders VERSION AS OF {version_to_check}")
display(df_version)
```
